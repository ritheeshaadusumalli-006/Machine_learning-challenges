# Final Challenge

Solve without looking at previous examples:

1. Find number of students.
2. Find number of features.
3. Calculate average marks.
4. Find highest marks.
5. Find students with marks >= 80.
6. Calculate average study hours.
7. Find the student with highest study hours.
8. Add 5 bonus marks to every student.
9. Reshape the dataset into another valid shape, if possible.
10. Transpose the dataset.

```python
# Final Challenge

students = np.array([
    [20, 5, 80],
    [21, 6, 75],
    [22, 8, 90],
    [23, 4, 65],
    [24, 7, 85]
])

# Solve all 10 tasks here.
num_students = students.shape[0]
print("Number of students:", num_students)
num_features = students.shape[1]
print("Number of features:", num_features)
avg_marks = students[:, 2].mean()
print("Average marks:", avg_marks)
highest_marks = students[:, 2].max()
print("Highest marks:", highest_marks)
top_students = students[students[:, 2] >= 80]
print("Students with marks >= 80:\n", top_students)
avg_study_hours = students[:, 1].mean()
print("Average study hours:", avg_study_hours)
top_study_student = students[students[:, 1] == students[:, 1].max()]
print("Student with highest study hours:\n", top_study_student)
students_with_bonus = students.copy()
students_with_bonus[:, 2] += 5
print("Students with bonus marks:\n", students_with_bonus)
reshaped = students.reshape(3, 5)
print("Reshaped dataset (3x5):\n", reshaped)
transposed = students.T
print("Transposed dataset:\n", transposed)
```
